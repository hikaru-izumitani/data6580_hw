| Aspect                             | SHAP                                                                                 | PDP (Partial Dependence Plot)                                                     | LIME                                                                        |
| ---------------------------------- | ------------------------------------------------------------------------------------ | --------------------------------------------------------------------------------- | --------------------------------------------------------------------------- |
| Primary purpose                    | Explain feature contributions to predictions                                         | Show how changing features affects average predictions                            | Approximate a model locally with a simpler, interpretable model             |
| Key question                       | Why did the model make this prediction?                                              | How does the model respond to changes in a feature?                               | What simpler model explains this prediction locally?                        |
| Explanation scope                  | Local and global                                                                     | Primarily global                                                                  | Local                                                                       |
| How it works                       | Calculates feature attributions based on Shapley values                              | Varies one or more features and averages predictions across observations          | Perturbs an observation and fits a weighted surrogate model                 |
| Model access required              | Depends on explainer; Kernel SHAP only needs a prediction function                   | Usually requires a callable prediction function                                   | Requires a callable prediction function                                     |
| Output                             | SHAP values for each feature                                                         | A curve or surface of average predictions                                         | Feature weights and rules from a local surrogate                            |
| Individual prediction explanations | Yes                                                                                  | Not with standard PDP                                                             | Yes                                                                         |
| Global feature importance          | Aggregate absolute SHAP values across observations                                   | Shows average feature-response patterns, not direct attribution rankings          | Requires aggregating many local explanations                                |
| Feature interactions               | Can capture interactions; SHAP interaction values are available for supported models | Standard PDP averages over other features; two-feature PDPs can show interactions | Can reflect local interactions to the extent captured by the surrogate      |
| Main strength                      | Additive feature attribution with a principled theoretical foundation                | Easy-to-understand feature-response visualization                                 | Flexible explanations for black-box models                                  |
| Main limitation                    | Correlated features and background-data choices affect interpretation                | May evaluate unrealistic feature combinations when features are correlated        | Sensitive to sampling, feature representation, and local surrogate fidelity |
| Computational cost                 | Low to high, depending on explainer and model                                        | Depends on number of grid points and dataset size                                 | Can be expensive due to repeated prediction queries                         |
| Best use case                      | Explaining individual predictions and measuring feature contributions                | Understanding overall model behavior and trends                                   | Explaining individual predictions when model internals are unavailable      |

Key takeaway: SHAP explains feature contributions, PDP explains average feature-response behavior, and LIME explains local behavior using a simpler surrogate model.

## 1. SHAP in one minute

SHAP assigns each feature a contribution to a particular model output.

For a given observation:

\[
\text{explained output} = \text{baseline output} + \sum_i \text{SHAP value}_i
\]

- **Baseline:** the explainer's expected model output for its chosen background/reference data.
- **SHAP value:** the feature's contribution relative to that baseline.
- **Positive contribution:** moves the explained output upward.
- **Negative contribution:** moves the explained output downward.

The phrase *explained output* matters. In regression, this is usually the predicted target. For classification, it may be a probability, log-odds, or another model output, depending on the explainer configuration. Always confirm the output space before interpreting values.

### Global vs. local explanations

- **Global importance:** commonly uses mean absolute SHAP values across observations. It measures the average magnitude of a feature's contribution, not whether the feature is inherently good or bad.
- **Local explanation:** shows how features contributed to one observation's prediction. A waterfall plot is useful for this.

## 2. A small helper for consistent plots

We'll use a helper to create readable plots. The examples below use tree-based random forests, so `shap.TreeExplainer` is a natural choice. For models that are not supported by TreeExplainer, consider the more general `shap.Explainer` API and select an appropriate masker/background dataset.

```python
def show_global_and_local(shap_values, X_explain, row_index=0, max_display=12):
    """Display global mean-|SHAP| importance and a local waterfall plot."""
    # Newer SHAP versions use Explanation objects for modern plotting.
    shap.plots.bar(shap_values, max_display=max_display)
    plt.show()

    shap.plots.waterfall(shap_values[row_index], max_display=max_display)
    plt.show()

    print("Observation index in explanation sample:", row_index)
    display(X_explain.iloc[[row_index]])
```

## 3. Regression example

We use scikit-learn's built-in diabetes dataset and a `RandomForestRegressor`.

The target is a continuous numeric measure from the dataset. This is only a demonstration of regression explainablity.

```python
diabetes = load_diabetes(as_frame=True)
X_reg = diabetes.data
y_reg = diabetes.target

X_reg_train, X_reg_test, y_reg_train, y_reg_test = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=42,
)

reg_model = RandomForestRegressor(
    n_estimators=200,
    max_depth=8,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1,
)
reg_model.fit(X_reg_train, y_reg_train)

reg_predictions = reg_model.predict(X_reg_test)

print(f"RMSE: {np.sqrt(mean_squared_error(y_reg_test, reg_predictions)):.3f}")
print(f"MAE:  {mean_absolute_error(y_reg_test, reg_predictions):.3f}")
print(f"R²:   {r2_score(y_reg_test, reg_predictions):.3f}")

display(pd.DataFrame({
    "actual": y_reg_test.iloc[:10].to_numpy(),
    "predicted": reg_predictions[:10],
}).round(2))
```

### 3.1 Compute regression SHAP values

`TreeExplainer` understands the structure of supported tree models and can compute contributions efficiently. We explain a small sample from the test set to keep plotting quick; increase `N_EXPLAIN` if you want broader global analysis.

```python
N_EXPLAIN = min(100, len(X_reg_test))
X_reg_explain = X_reg_test.iloc[:N_EXPLAIN].copy()

reg_explainer = shap.TreeExplainer(reg_model)
reg_shap_values = reg_explainer(X_reg_explain)

print("SHAP values shape:", reg_shap_values.values.shape)
print("Expected/baseline value:", reg_explainer.expected_value)
```

### 3.2 Global importance and one local explanation

The bar plot ranks features by mean absolute SHAP value across the explanation sample. The waterfall plot explains the first observation in that sample.

In the waterfall plot:
- `E[f(X)]` is the baseline.
- Red features push the prediction higher.
- Blue features push it lower.
- The final value is the model output being explained.

```python
show_global_and_local(
    reg_shap_values,
    X_reg_explain,
    row_index=0,
    max_display=12,
)
```

### 3.3 Check additivity

For a regression explanation, the sum of the baseline and SHAP values should reconstruct the prediction, up to small floating-point differences.

This is a useful sanity check to add to automated tests.

```python
row = 0
reconstructed = (
    np.asarray(reg_shap_values.base_values[row]).item()
    + reg_shap_values.values[row].sum()
)
model_prediction = reg_model.predict(X_reg_explain.iloc[[row]])[0]

print(f"SHAP reconstructed output: {reconstructed:.8f}")
print(f"Model prediction:           {model_prediction:.8f}")
print(f"Absolute difference:        {abs(reconstructed - model_prediction):.3e}")

assert np.isclose(reconstructed, model_prediction, atol=1e-5), (
    "SHAP additivity check failed. Check SHAP/model versions and output settings."
)
```

### 3.4 Inspect individual contributions as a table

This table sorts features by absolute contribution for one observation. The sign shows direction; the magnitude shows contribution size in the model's explained output units.

```python
row = 0
local_reg = pd.DataFrame({
    "feature": X_reg_explain.columns,
    "feature_value": X_reg_explain.iloc[row].to_numpy(),
    "shap_value": reg_shap_values.values[row],
})
local_reg["abs_shap"] = local_reg["shap_value"].abs()
display(local_reg.sort_values("abs_shap", ascending=False).drop(columns="abs_shap").head(12))
```

## 4. Binary classification example

We use the breast cancer dataset, which has two target classes. The model predicts a class label and can also produce class probabilities with `predict_proba`.

**Interpretation caution:** the class-specific SHAP output scale depends on the explainer and SHAP version. Do not assume a SHAP value is a percentage-point change in probability. We will inspect the SHAP output and compare it with model probabilities.

```python
cancer = load_breast_cancer(as_frame=True)
X_bin = cancer.data
y_bin = cancer.target

X_bin_train, X_bin_test, y_bin_train, y_bin_test = train_test_split(
    X_bin,
    y_bin,
    test_size=0.20,
    random_state=42,
    stratify=y_bin,
)

bin_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=8,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1,
)
bin_model.fit(X_bin_train, y_bin_train)

bin_predictions = bin_model.predict(X_bin_test)
bin_probabilities = bin_model.predict_proba(X_bin_test)

print("Classes:", list(bin_model.classes_))
print(f"Accuracy: {accuracy_score(y_bin_test, bin_predictions):.3f}")
print(classification_report(
    y_bin_test,
    bin_predictions,
    target_names=list(cancer.target_names),
))

ConfusionMatrixDisplay.from_predictions(
    y_bin_test,
    bin_predictions,
    display_labels=cancer.target_names,
    cmap=None,
)
plt.title("Binary classifier confusion matrix")
plt.show()
```

### 4.1 Compute binary-classification SHAP values

Modern SHAP versions generally return an `Explanation` object from calling the explainer. For a binary classifier, the returned values may have shape `(observations, features)` or `(observations, features, classes)`, depending on the model/explainer/version. The following cell prints the shape before we select the class output.

```python
N_BIN_EXPLAIN = min(100, len(X_bin_test))
X_bin_explain = X_bin_test.iloc[:N_BIN_EXPLAIN].copy()

bin_explainer = shap.TreeExplainer(bin_model)
bin_shap_all = bin_explainer(X_bin_explain)

print("SHAP values shape:", bin_shap_all.values.shape)
print("Base values shape:", np.asarray(bin_shap_all.base_values).shape)
print("Model classes:", list(bin_model.classes_))
```

### 4.2 Select and plot the explanation for one class

We'll explain `bin_model.classes_[1]`, the second class in the model's class ordering. Always use `model.classes_` to map an output index to its label; don't assume class `1` means a particular business outcome without checking the dataset.

This cell handles the common SHAP output shapes. If a particular SHAP/model version uses an unexpected shape or output space, it raises an informative error instead of silently making an incorrect interpretation.

```python
def select_class_explanation(shap_output, class_index, model, X_explain):
    """Return a 2D SHAP Explanation for one class, where supported."""
    values = np.asarray(shap_output.values)
    base = np.asarray(shap_output.base_values)
    n_classes = len(model.classes_)

    if values.ndim == 2:
        # A 2D result explains one output. This is common for binary
        # tree classifiers, but its output scale must be verified.
        selected_values = values
        if base.ndim == 0:
            selected_base = np.full(len(X_explain), base.item())
        elif base.ndim == 1 and len(base) == len(X_explain):
            selected_base = base
        else:
            raise ValueError(f"Unexpected 2D-output base shape: {base.shape}")

    elif values.ndim == 3:
        # SHAP commonly uses (samples, features, outputs/classes).
        if values.shape[2] != n_classes:
            raise ValueError(
                f"Last SHAP dimension {values.shape[2]} does not match "
                f"number of classes {n_classes}; shape={values.shape}"
            )
        selected_values = values[:, :, class_index]

        if base.ndim == 0:
            selected_base = np.full(len(X_explain), base.item())
        elif base.ndim == 1 and len(base) == n_classes:
            selected_base = np.full(len(X_explain), base[class_index])
        elif base.ndim == 2 and base.shape == (len(X_explain), n_classes):
            selected_base = base[:, class_index]
        else:
            raise ValueError(f"Unexpected class-output base shape: {base.shape}")
    else:
        raise ValueError(f"Unsupported SHAP values shape: {values.shape}")

    return shap.Explanation(
        values=selected_values,
        base_values=selected_base,
        data=X_explain.to_numpy(),
        feature_names=list(X_explain.columns),
    )


binary_class_index = 1
binary_class_label = bin_model.classes_[binary_class_index]
bin_class_shap = select_class_explanation(
    bin_shap_all,
    binary_class_index,
    bin_model,
    X_bin_explain,
)

print("Explained class label:", binary_class_label)
show_global_and_local(
    bin_class_shap,
    X_bin_explain,
    row_index=0,
    max_display=12,
)
```

### 4.3 Compare prediction probability and SHAP output

This is a comparison, not an assumption that the SHAP sum equals the class probability. If the explainer is using a raw margin/log-odds output, compare the SHAP reconstruction to that raw output instead. The next cell prints the model probability and SHAP reconstruction side by side to make the scale issue visible.

```python
row = 0
shap_reconstructed = (
    np.asarray(bin_class_shap.base_values[row]).item()
    + bin_class_shap.values[row].sum()
)
class_probability = bin_model.predict_proba(
    X_bin_explain.iloc[[row]]
)[0, binary_class_index]

print("Class label:", binary_class_label)
print(f"predict_proba for selected class: {class_probability:.6f}")
print(f"SHAP baseline + contributions:   {shap_reconstructed:.6f}")
print(
    "These numbers are directly comparable only if the SHAP explainer "
    "is configured to explain the same output scale as predict_proba."
)
```

### 4.4 Optional: request probability-space explanations

For tree models, probability-space explanations can be useful when you want contributions in probability units. Support depends on the model and SHAP version; interventional perturbation generally requires a background dataset. If this cell errors for your installed versions, keep the default TreeExplainer result and interpret its output scale accordingly.

```python
# Optional probability-space explainer. May be slower than the default.
try:
    background = X_bin_train.sample(
        n=min(100, len(X_bin_train)),
        random_state=42,
    )
    bin_probability_explainer = shap.TreeExplainer(
        bin_model,
        data=background,
        feature_perturbation="interventional",
        model_output="probability",
    )
    bin_probability_shap = bin_probability_explainer(X_bin_explain)

    print("Probability-space SHAP shape:", bin_probability_shap.values.shape)
    print("Probability-space baseline shape:",
          np.asarray(bin_probability_shap.base_values).shape)

    # For binary classifier versions returning 3D values, select class 1.
    bin_probability_class_shap = select_class_explanation(
        bin_probability_shap,
        binary_class_index,
        bin_model,
        X_bin_explain,
    )
    show_global_and_local(
        bin_probability_class_shap,
        X_bin_explain,
        row_index=0,
        max_display=12,
    )
except Exception as exc:
    print("Probability-space explanation was not available in this environment.")
    print(type(exc).__name__ + ":", exc)
```

## 5. Multiclass classification example

We use the wine dataset, which has three classes. Multiclass SHAP explains each class output separately. A feature can support one class while opposing another, so always inspect which class is being explained.

```python
wine = load_wine(as_frame=True)
X_multi = wine.data
y_multi = wine.target

X_multi_train, X_multi_test, y_multi_train, y_multi_test = train_test_split(
    X_multi,
    y_multi,
    test_size=0.20,
    random_state=42,
    stratify=y_multi,
)

multi_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=8,
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1,
)
multi_model.fit(X_multi_train, y_multi_train)

multi_predictions = multi_model.predict(X_multi_test)
multi_probabilities = multi_model.predict_proba(X_multi_test)

print("Class labels:", list(multi_model.classes_))
print("Class names:", list(wine.target_names))
print(f"Accuracy: {accuracy_score(y_multi_test, multi_predictions):.3f}")
print(classification_report(
    y_multi_test,
    multi_predictions,
    target_names=list(wine.target_names),
))

ConfusionMatrixDisplay.from_predictions(
    y_multi_test,
    multi_predictions,
    display_labels=wine.target_names,
)
plt.title("Multiclass classifier confusion matrix")
plt.show()
```

### 5.1 Compute multiclass SHAP values

```python
N_MULTI_EXPLAIN = min(100, len(X_multi_test))
X_multi_explain = X_multi_test.iloc[:N_MULTI_EXPLAIN].copy()

multi_explainer = shap.TreeExplainer(multi_model)
multi_shap_all = multi_explainer(X_multi_explain)

print("SHAP values shape:", multi_shap_all.values.shape)
print("Base values shape:", np.asarray(multi_shap_all.base_values).shape)
print("Model class order:", list(multi_model.classes_))
```

### 5.2 Explain each class

We build a dictionary mapping each actual class label to its own `Explanation`. Each class gets a global importance plot and a local waterfall plot for the first observation.

For modern SHAP versions, multiclass values usually have shape `(observations, features, classes)`. The helper also supports a 2D explanation where a model explains only one output.

```python
multi_class_explanations = {}

for class_index, class_label in enumerate(multi_model.classes_):
    class_exp = select_class_explanation(
        multi_shap_all,
        class_index,
        multi_model,
        X_multi_explain,
    )
    multi_class_explanations[class_label] = class_exp

    class_name = wine.target_names[int(class_label)]
    print(f"Class index={class_index}, label={class_label}, name={class_name}")

    show_global_and_local(
        class_exp,
        X_multi_explain,
        row_index=0,
        max_display=10,
    )
```

### 5.3 Compare the same observation across all classes

The next table shows each class's predicted probability for the first explained observation. The following plots show the class-specific SHAP contributions for that same observation.

Remember: each class's SHAP values explain that class's output. Do not add contributions from different classes together.

```python
row = 0
probability_row = multi_model.predict_proba(X_multi_explain.iloc[[row]])[0]

display(pd.DataFrame({
    "class_index": np.arange(len(multi_model.classes_)),
    "class_label": multi_model.classes_,
    "class_name": [wine.target_names[int(c)] for c in multi_model.classes_],
    "predicted_probability": probability_row,
}).sort_values("predicted_probability", ascending=False).reset_index(drop=True))

for class_index, class_label in enumerate(multi_model.classes_):
    exp = multi_class_explanations[class_label]
    print(f"SHAP contributions for class {wine.target_names[int(class_label)]}")
    shap.plots.waterfall(exp[row], max_display=10)
    plt.show()
```


## 6. Model-agnostic example with `shap.KernelExplainer`

`TreeExplainer` uses the structure of tree models and is usually much faster for supported tree ensembles. **`KernelExplainer` is model-agnostic**: it can explain many models as long as you provide a callable that maps a batch of feature rows to model outputs.

Kernel SHAP estimates contributions by evaluating the model on combinations of present and masked features. This flexibility can be expensive, so we will:
1. Use a small background sample to represent the reference population.
2. Explain only a few test rows.
3. Limit `nsamples` to keep this demonstration practical.

The example below uses the regression model trained earlier. The callable returns predictions in the same units as the regression target, so the SHAP baseline and contributions are interpretable in those units. KernelExplainer is useful for models that TreeExplainer does not directly support, though runtime can grow quickly with feature count and sample count.


```python

# Keep KernelExplainer small for a quick notebook demonstration.
# Background rows define the reference distribution for the explanation.
X_kernel_background = shap.sample(X_reg_train, 20, random_state=42)
X_kernel_explain = X_reg_test.iloc[:5].copy()

# KernelExplainer accepts a prediction function and a background dataset.
# DataFrames are converted to NumPy arrays inside the wrapper for consistency.
def regression_predict_for_kernel(data):
    data_frame = pd.DataFrame(data, columns=X_reg_train.columns)
    return reg_model.predict(data_frame)

kernel_explainer = shap.KernelExplainer(
    regression_predict_for_kernel,
    X_kernel_background,
)

# nsamples controls the number of coalition samples used to estimate SHAP values.
# More samples may improve stability but increase runtime.
kernel_shap_values = kernel_explainer.shap_values(
    X_kernel_explain,
    nsamples=100,
)

kernel_shap_values = np.asarray(kernel_shap_values)
print("Kernel SHAP values shape:", kernel_shap_values.shape)
print("Kernel SHAP expected value:", kernel_explainer.expected_value)
```


### 6.1 Plot the Kernel SHAP explanation

We wrap the computed values in a modern `shap.Explanation` object so the same plotting API can be used. The waterfall plot explains the first of the five selected observations.


```python

kernel_base = np.asarray(kernel_explainer.expected_value).reshape(-1)[0]
kernel_explanation = shap.Explanation(
    values=kernel_shap_values,
    base_values=np.full(len(X_kernel_explain), kernel_base),
    data=X_kernel_explain.to_numpy(),
    feature_names=list(X_kernel_explain.columns),
)

show_global_and_local(
    kernel_explanation,
    X_kernel_explain,
    row_index=0,
    max_display=12,
)
```


### 6.2 Validate the Kernel SHAP reconstruction

Kernel SHAP is an estimate, so the reconstruction may differ slightly from the model prediction. The discrepancy should generally be small enough for this demonstration; if it is larger than expected, increase `nsamples` and compare again.

Because the model callable returns regression predictions, the sum of the baseline and contributions should approximate the predicted target.


```python

kernel_row = 0
kernel_reconstructed = (
    kernel_explanation.base_values[kernel_row]
    + kernel_explanation.values[kernel_row].sum()
)
kernel_prediction = reg_model.predict(X_kernel_explain.iloc[[kernel_row]])[0]

print(f"Kernel SHAP reconstructed output: {kernel_reconstructed:.8f}")
print(f"Model prediction:                  {kernel_prediction:.8f}")
print(f"Absolute difference:               {abs(kernel_reconstructed - kernel_prediction):.3e}")
```


### When should you use `KernelExplainer`?

| Situation | Recommended starting point |
|---|---|
| Supported tree model and many rows | `shap.TreeExplainer` |
| Model-agnostic explanation for a small number of rows | `shap.KernelExplainer` |
| General SHAP interface for many model types | `shap.Explainer`, with an appropriate masker/model callable |

**Practical notes**
- Kernel SHAP can be much slower than Tree SHAP, especially when there are many features.
- The background dataset affects the baseline and the interpretation of contributions. Select representative background rows.
- Consider scaling or grouping features when the model has a large feature space, but preserve the model's expected input format.
- For classification, the callable can return `predict_proba` if you specifically want to explain class probabilities. In that case, explain one output column at a time or verify the output shape and baseline carefully.
- Kernel SHAP explanations are not causal explanations.


## 7. Reusable utility: summarize SHAP importance

For reporting, it is often useful to export mean absolute SHAP values to a DataFrame. This helper works with a single-output `Explanation`. For multiclass, call it once per class.

```python
def shap_importance_table(explanation):
    """Rank features by mean absolute SHAP value across observations."""
    values = np.asarray(explanation.values)
    if values.ndim != 2:
        raise ValueError(
            "Pass a single-output Explanation with shape (samples, features)."
        )

    result = pd.DataFrame({
        "feature": explanation.feature_names,
        "mean_abs_shap": np.abs(values).mean(axis=0),
        "mean_shap": values.mean(axis=0),
    })
    return result.sort_values("mean_abs_shap", ascending=False).reset_index(drop=True)


print("Regression importance")
display(shap_importance_table(reg_shap_values).head(15))

print("Binary class importance")
display(shap_importance_table(bin_class_shap).head(15))

print("Multiclass class 0 importance")
display(shap_importance_table(multi_class_explanations[multi_model.classes_[0]]).head(15))
```

## 9. Adapting this notebook to your own model

### Replace the example data
- Set `X` to a pandas DataFrame of model input features.
- Set `y` to the target.
- Keep feature names in the DataFrame; they make explanations much easier to interpret.
- Use a train/test split that matches your use case (for example, a time-based split for time-series or temporal data).

### Replace the model
- Random forest / supported tree ensembles: `shap.TreeExplainer(model)` is a good starting point.
- Other estimators: try `shap.Explainer(model, background_data)` or a model-agnostic explainer appropriate to the estimator. Runtime and explanation semantics vary.
- For a pipeline, ensure the explainer sees the same transformed feature representation the model uses, and preserve a mapping back to human-readable feature names.

### Production checklist
1. Pin SHAP, NumPy, scikit-learn, and model-library versions.
2. Record model version, feature schema, class order, baseline/background data version, and output scale.
3. Test SHAP additivity for representative inputs.
4. Handle missing values and categorical encodings consistently with model inference.
5. Treat SHAP explanations as diagnostic evidence, not causal proof.
6. Avoid logging sensitive raw features unless your data policy permits it.
7. For regulated decisions, validate explanation stability and document limitations.

## 10. Troubleshooting

- **Import / binary compatibility errors:** create a clean virtual environment and reinstall dependencies.
- **Plot does not render:** run `%matplotlib inline` in a notebook cell and ensure the notebook frontend supports inline plots.
- **Unexpected SHAP shape:** print `explanation.values.shape`, `explanation.base_values.shape`, and `model.classes_`; class ordering and return shapes can vary by SHAP version and model.
- **Additivity mismatch:** confirm the explained model output scale, check preprocessing, and avoid comparing raw-output SHAP sums to probabilities.
- **Slow explanations:** reduce `N_EXPLAIN`, reduce background sample size, or compute explanations only for selected records.

## 10. Black-box SHAP with `KernelExplainer` when you cannot access the model

This is a different use case from the earlier tree-based examples.

**Assumption:** you cannot inspect or load the model itself. You can only submit input records to a prediction service or other callable system and receive its outputs. For example, you may have access to a REST endpoint, an internal prediction API, a vendor-hosted model, or a deployed model whose weights and implementation are unavailable.

`shap.KernelExplainer` can work in this situation because it needs a **prediction function**, not the model's internal structure. The function you provide must accept a batch of input rows and return the corresponding output(s).

Conceptually:

1. Select a background dataset that represents typical/reference inputs.
2. Wrap the prediction API so it accepts a batch of rows and returns predictions in the format SHAP expects.
3. Pass that wrapper and the background rows to `shap.KernelExplainer`.
4. Compute SHAP values for the rows you want to explain.

**Important limitation:** you still need a way to query the system for predictions. If you have only a static input/output table and cannot query the model on perturbed inputs, ordinary Kernel SHAP generally cannot identify the missing coalition outputs. You may explain the observed data with other statistical methods, but that is not the same as obtaining faithful model-specific SHAP values.

### 10.1 Create a black-box prediction API for demonstration

To make this notebook runnable without credentials or a real service, we'll simulate a prediction endpoint using a local function. In your real use case, replace `black_box_predict()` with an API client call. The KernelExplainer code below does not inspect the model used behind that function.

The local function is only a stand-in for a remote black box: the explanation code sees the input/output interface.

```python
# Demo-only black-box service.
# Imagine the implementation lives on a remote server and is not available to you.
# In a real deployment, replace the body with an API request.

def black_box_predict(input_rows: pd.DataFrame) -> np.ndarray:
    """Return one regression prediction per input row.

    Contract:
      - input_rows: DataFrame with the expected input columns
      - return: 1D numeric array with one output per row
    """
    # Simulated service call. The "hidden" model is used only to make this
    # notebook executable end-to-end; the SHAP wrapper below never accesses it.
    return reg_model.predict(input_rows)


# A realistic background sample and observations to explain.
# These are inputs only; we call the prediction interface for outputs.
X_blackbox_background = X_reg_train.sample(
    n=min(20, len(X_reg_train)),
    random_state=42,
)
X_blackbox_explain = X_reg_test.iloc[:5].copy()

blackbox_outputs = black_box_predict(X_blackbox_explain)
print("Inputs sent to black box:")
display(X_blackbox_explain.head())
print("Outputs received:")
display(pd.DataFrame({
    "prediction_from_black_box": blackbox_outputs
}, index=X_blackbox_explain.index))
```

### 10.2 Build a robust prediction wrapper

`KernelExplainer` may pass a NumPy array to the function, not a DataFrame. The wrapper restores the expected column names and ordering, calls the black-box interface, and returns a numeric array.

For a real HTTP API, also handle authentication, timeouts, rate limits, retries, schema validation, and service errors. Keep requests batched where the endpoint supports batch inference; Kernel SHAP can require many calls.

```python
BLACKBOX_FEATURES = list(X_blackbox_background.columns)

def blackbox_predict_for_shap(rows):
    """Adapt SHAP's array input to the black-box input/output contract."""
    rows_df = pd.DataFrame(rows, columns=BLACKBOX_FEATURES)
    outputs = black_box_predict(rows_df)

    outputs = np.asarray(outputs, dtype=float)

    # For a single-output regression API, normalize to shape (n_rows,).
    if outputs.ndim == 2 and outputs.shape[1] == 1:
        outputs = outputs[:, 0]

    if outputs.ndim != 1 or len(outputs) != len(rows_df):
        raise ValueError(
            "Expected one scalar output per input row. "
            f"Received output shape {outputs.shape} for {len(rows_df)} rows."
        )

    return outputs
```

### 10.3 Compute Kernel SHAP values without model access

Kernel SHAP approximates Shapley values by querying the prediction function on synthetic combinations of observed and background feature values. It is model-agnostic, but can be significantly slower than `TreeExplainer`.

- `background`: reference records used to represent “missing” features.
- `nsamples`: approximation budget per explained row. Larger values usually cost more time.
- Start with a small explanation sample and background sample, then increase them if needed.

Because this example uses a regression API, each explained row receives one SHAP value per input feature.

```python
kernel_explainer = shap.KernelExplainer(
    blackbox_predict_for_shap,
    data=X_blackbox_background,
)

# A modest sample for a runnable demonstration.
kernel_shap_values = kernel_explainer.shap_values(
    X_blackbox_explain,
    nsamples=100,
)

kernel_shap_values = np.asarray(kernel_shap_values)

# Older SHAP versions may return a list for multi-output models; this
# particular API contract is single-output regression.
if kernel_shap_values.ndim == 3 and kernel_shap_values.shape[-1] == 1:
    kernel_shap_values = kernel_shap_values[:, :, 0]

print("Kernel SHAP values shape:", kernel_shap_values.shape)
print("Kernel SHAP expected value:", kernel_explainer.expected_value)
```

### 10.4 Visualize the black-box explanation

Create a modern `shap.Explanation` object from the Kernel SHAP values. The waterfall plot explains one record; the bar plot ranks features by mean absolute contribution across the five records.

```python
kernel_base = np.asarray(kernel_explainer.expected_value).reshape(-1)
if kernel_base.size != 1:
    raise ValueError(
        "This example expects a single-output regression API. "
        "For multi-output APIs, create one explanation per output."
    )

kernel_explanation = shap.Explanation(
    values=kernel_shap_values,
    base_values=np.full(len(X_blackbox_explain), kernel_base.item()),
    data=X_blackbox_explain.to_numpy(),
    feature_names=list(X_blackbox_explain.columns),
)

shap.plots.bar(kernel_explanation, max_display=12)
plt.show()

shap.plots.waterfall(kernel_explanation[0], max_display=12)
plt.show()
```
